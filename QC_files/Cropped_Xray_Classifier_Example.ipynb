## Deploying the cropped full body transparent image classifier

```python
import pandas as pd                         # used to load the CSV
from fastai.vision.all import *             # needed for load_learner and predict
from fastai.medical.imaging import PILDicom # to load DICOM images
```

```python
# Define these two functions exactly the same way
# These were defined during training and need to be established for the model to load
def get_x(label_df):
    return label_df['filename']

def get_y(label_df):
    return label_df['label']
```

```python
# Load in pandas dataframe with the full path name of your .dcm images in one of your columns
df = pd.read_csv("path/to/full_path_name_of_images")
```

```python
# Load pre-trained model
learn = load_learner('Cropped_classifier_updated_model.pkl')
```

```python
# Use PILDicom to load dicom images from the path name and predict 'crop' versus 'no_crop'

# Example loop to loop through dataframe where "filename" is a column with 
# the path name of .dcm images
for image in df["filename"][0:50]:
    print(image)
    dicom_img = PILDicom.create(image)
    prediction = learn.predict(dicom_img)[0]
    print(prediction)
```
